```python
import pandas as pd

# (이전 코드에서 이어짐)
# trend_df 에는 ['keyword', 'year_month', 'view_count'] 가 있는 상태입니다.

# ==========================================
# 🌟 STEP 1: X축 정규화 (상대적 시간축 '경과 월' 만들기)
# ==========================================
# 1. 각 키워드별로 가장 처음 영상이 올라온(또는 검색된) 달을 찾습니다. (이게 T=0 기준점이 됩니다)
start_months = trend_df.groupby('keyword')['year_month'].transform('min')

# 2. 현재 월과 최초 등장 월의 차이를 계산하여 '경과_월(Month_T)' 컬럼을 만듭니다.
# (연도 차이 * 12) + (월 차이) = 총 몇 개월이 지났는지 나옴
trend_df['경과_월'] = (trend_df['year_month'].dt.year - start_months.dt.year) * 12 + \
                      (trend_df['year_month'].dt.month - start_months.dt.month)

# ==========================================
# 🌟 STEP 2: Y축 정규화 (조회수 스케일 맞추기)
# ==========================================
# 아이템마다 조회수 스케일(100만 vs 1000만)이 다르므로 패턴만 보기 위해 최고점을 100으로 누릅니다.
# 각 키워드별 가장 높았던 조회수를 찾습니다.
max_views = trend_df.groupby('keyword')['view_count'].transform('max')

# 최고점 대비 현재 월의 조회수 비율(%)을 계산합니다.
trend_df['조회수_정규화(%)'] = (trend_df['view_count'] / max_views) * 100


# ==========================================
# 🌟 STEP 3: 정규화된 데이터로 그래프 다시 그리기
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# X축은 '경과_월', Y축은 '조회수_정규화(%)' 를 사용합니다!
sns.lineplot(data=trend_df, x='경과_월', y='조회수_정규화(%)', hue='keyword', marker='o', linewidth=2)

plt.title('아이템별 유행 수명주기 비교 (T=0 정규화)', fontsize=16)
plt.xlabel('최초 등장 이후 경과된 개월 수 (T+n)', fontsize=12) # X축 이름 변경
plt.ylabel('최고점 대비 조회수 비율 (%)', fontsize=12)         # Y축 이름 변경
plt.grid(True, linestyle='--', alpha=0.6)

# X축 눈금을 0, 1, 2, 3.. 처럼 딱 떨어지게 설정
max_month = trend_df['경과_월'].max()
plt.xticks(range(0, max_month + 1))

plt.tight_layout()
plt.show()
```

```python
import time
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from googleapiclient.discovery import build

# 한글 폰트를 아예 안 쓰므로 기본 폰트 사용 (폰트 깨짐 이슈 완벽 해결!)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.unicode_minus'] = False

# ==========================================
# 1. API 세팅 및 검색어 리스트
# ==========================================
API_KEY = "AIzaSyAiIOgmEPQMYUp0xRpyqkUWxQ_hO9GDgKI"
youtube = build('youtube', 'v3', developerKey=API_KEY)

# 기존 데이터프레임에서 리스트로 가져왔다고 가정
keyword_list = ["두바이 초콜릿", "탕후루", "요아정", "마라탕", "흑당버블티"]
video_data = []

print("데이터 수집을 시작합니다...")

# ==========================================
# 2. 데이터 수집 (한글 이름 대신 인덱스 번호 부여!)
# ==========================================
# enumerate를 사용해 0, 1, 2... 인덱스 번호를 같이 뽑아냅니다.
for idx, kw in enumerate(keyword_list):
    # 한글 키워드 대신 Item_0, Item_1 형태로 라벨링
    item_label = f"Item_{idx}"

    next_page_token = None
    print(f"👉 [{item_label}] 수집 중... (원래 키워드: {kw})")

    for i in range(1): # 3페이지씩만 테스트
        try:
            res = youtube.search().list(
                q=kw, part='snippet', maxResults=50, type='video',
                order='viewCount', pageToken=next_page_token,
                publishedAfter='2022-01-01T00:00:00Z'
            ).execute()

            for item in res['items']:
                v_id = item['id']['videoId']
                snippet = item['snippet']

                try:
                    stats_res = youtube.videos().list(part='statistics', id=v_id).execute()
                    stats = stats_res['items'][0]['statistics']

                    video_data.append({
                        'item_id': item_label,  # 🔥 한글(kw) 대신 인덱스 기반 라벨(Item_n) 저장
                        'video_id': v_id,
                        'published_at': snippet['publishedAt'],
                        'view_count': int(stats.get('viewCount', 0))
                    })
                except:
                    continue

            next_page_token = res.get('nextPageToken')
            if not next_page_token: break

        except Exception as e:
            print(f"API 에러: {e}")
            break

    time.sleep(1) # API 할당량 보호

df_raw = pd.DataFrame(video_data)
print(f"총 {len(df_raw)}개의 영상 데이터 수집 완료!")


# ==========================================
# 3. 데이터 전처리 (월별 그룹핑 -> T=0 정렬 -> Y축 정규화 -> 이동평균)
# ==========================================
df_raw['published_at'] = pd.to_datetime(df_raw['published_at'])
df_raw['year_month'] = df_raw['published_at'].dt.to_period('M').dt.to_timestamp()

# 1) 월별 총 조회수로 뭉치기
trend_df = df_raw.groupby(['item_id', 'year_month'])['view_count'].sum().reset_index()

# 2) T=0 정렬 (상대 시간축)
start_months = trend_df.groupby('item_id')['year_month'].transform('min')
trend_df['month_t'] = (trend_df['year_month'].dt.year - start_months.dt.year) * 12 + \
                      (trend_df['year_month'].dt.month - start_months.dt.month)

# 3) Y축 정규화 (최고점 100%)
max_views = trend_df.groupby('item_id')['view_count'].transform('max')
trend_df['view_ratio(%)'] = (trend_df['view_count'] / max_views) * 100

# 4) 🌟 선 부드럽게 만들기 (3개월 이동평균 스무딩)
trend_df['smoothed_ratio(%)'] = trend_df.groupby('item_id')['view_ratio(%)'].transform(
    lambda x: x.rolling(window=3, min_periods=1).mean()
)


# ==========================================
# 4. 시각화 (한글 없이 영문으로만 깔끔하게 출력)
# ==========================================
plt.figure(figsize=(10, 6))

# 노이즈가 제거된 'smoothed_ratio(%)'를 Y축으로 사용합니다.
sns.lineplot(
    data=trend_df,
    x='month_t',
    y='smoothed_ratio(%)',
    hue='item_id',
    marker='o',
    linewidth=2,
    alpha=0.7 # 선이 겹쳐도 잘 보이게 투명도 약간 추가
)

plt.title('Trend Lifecycle Pattern Verification (T=0, Smoothed)', fontsize=16)
plt.xlabel('Months since first appearance (T+n)', fontsize=12)
plt.ylabel('View Count Ratio (% to Max)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# X축 촘촘하게 설정 (최대 36개월까지만 잘라서 보여주기)
max_month = min(trend_df['month_t'].max(), 36)
plt.xlim(-1, max_month + 1)
plt.xticks(range(0, max_month + 1, 2)) # 2개월 단위로 눈금

plt.legend(title='Item Index', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

수집된 데이터를 '주차별(Week)'로 묶고, 분석하기 좋게 다듬는 단계입니다.

```python
# --- [Cell 2: 정제 - 단위 시간당 반응(Velocity) 기반 주차별 그룹화] ---
import pandas as pd

# 1. 날짜 데이터 타입 변환 (현재 시간 기준)
df_raw['published_at'] = pd.to_datetime(df_raw['published_at'])
now = pd.to_datetime('now', utc=True)

# 2. 영상 생존 일수(나이) 및 '하루 평균 조회수(가속도)' 추출 ★누적의 함정 해결★
df_raw['video_age_days'] = (now - df_raw['published_at']).dt.days
df_raw['video_age_days'] = df_raw['video_age_days'].replace(0, 1) # 오늘 올라온 영상 0 나누기 에러 방지
df_raw['views_per_day'] = df_raw['view_count'] / df_raw['video_age_days']

# 3. 주차별(Week) 그룹화
df_raw['week'] = df_raw['published_at'].dt.to_period('W').apply(lambda r: r.start_time)

# 누적합이 아닌 해당 주차 영상들의 '평균 가속도' 산출
df_weekly = df_raw.groupby('week').agg(
    video_supply=('video_id', 'count'),          # 공급량: 새로 올라온 영상 개수
    avg_velocity=('views_per_day', 'mean'),      # 반응량: 하루 평균 조회수
    total_comments=('comment_count', 'sum')      # 향후 댓글 분석을 위한 보존
).reset_index()

# 4. 노이즈 제거 (영상이 1개 이상 올라온 주차만 분석)
df_weekly = df_weekly[df_weekly['video_supply'] > 0].reset_index(drop=True)

print("--- Cell 2: 단위 시간당 반응을 반영한 데이터 정제 완료! ---")
display(df_weekly.head())
```

'파생 변수'를 만들고, 모든 단위를 비교할 수 있게 Z-score로 스케일링하는 핵심 단계입니다.

```python
# --- [Cell 3: 분석 - 파생 변수 생성 및 Z-Score 스케일링] ---

# 1. 수요-공급 압력 (Demand Pressure) 계산
# 하루 평균 조회수를 공급량(영상 수)으로 나누어, 영상 1개가 감당하는 파괴력 계산
df_weekly['demand_pressure'] = df_weekly['avg_velocity'] / df_weekly['video_supply']

# 2. Z-Score 스케일링 함수 정의 (평균 0, 표준편차 1)
def calculate_zscore(col):
    return (col - col.mean()) / col.std()

# 3. 주요 지표들에 Z-Score 적용
columns_to_scale = ['video_supply', 'avg_velocity', 'demand_pressure']
for col in columns_to_scale:
    df_weekly[f'{col}_Z'] = calculate_zscore(df_weekly[col])

print("--- Cell 3: 파생 변수 생성 및 Z-Score 스케일링 완료! ---")
display(df_weekly[['week', 'video_supply_Z', 'avg_velocity_Z', 'demand_pressure_Z']].head())
```

Z-score로 통일된 지표들을 한눈에 보고 유행의 '임계점'을 포착하는 단계입니다.

```python
# --- [Cell 4: 유행 임계점 시각화 (Velocity 반영)] ---
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 7))

# 1. 시각화할 Z-Score 라인 그리기 (변수명 업데이트)
sns.lineplot(data=df_weekly, x='week', y='video_supply_Z', marker='o', label='Supply (Video Count)', color='grey', alpha=0.6)
sns.lineplot(data=df_weekly, x='week', y='avg_velocity_Z', marker='s', label='Demand (Daily Velocity)', color='blue') # 여기 수정됨
sns.lineplot(data=df_weekly, x='week', y='demand_pressure_Z', marker='^', label='Pressure (Velocity per Video)', color='skyblue', linewidth=2.5)

# 2. 기준선(Baseline) 긋기
plt.axhline(0, color='black', linestyle='--', linewidth=1) # 평균선
plt.axhline(1.5, color='orange', linestyle=':', linewidth=2, label='Breakout Threshold (Z=1.5)') # 유행 임계점

# 3. 그래프 꾸미기
plt.title("Trend Analysis: Breakout Detection (Velocity Z-Score)", fontsize=16)
plt.ylabel("Z-Score (Standardized)", fontsize=12)
plt.xlabel("Upload Week", fontsize=12)
plt.xticks(rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
import requests
import pandas as pd
import os

# ==========================================
# 1. API 키 세팅 (실무에서는 .env 파일이나 환경변수로 숨깁니다!)
# ==========================================
# os.getenv("NAVER_CLIENT_ID") 방식으로 불러오는 것이 정석입니다.
CLIENT_ID = "799akmjBLWabHq1xjD8W"
CLIENT_SECRET = "z2U_2bGCh5"

# ==========================================
# 2. 다중 트렌드 데이터 조회 파이프라인 함수
# ==========================================
def get_naver_trend_pipeline(keywords: list, start_date: str, end_date: str, time_unit='date'):
    url = "https://openapi.naver.com/v1/datalab/search"

    headers = {
        "X-Naver-Client-Id": CLIENT_ID,
        "X-Naver-Client-Secret": CLIENT_SECRET,
        "Content-Type": "application/json"
    }

    # 🌟 [업그레이드 1] 입력받은 키워드 리스트를 API가 원하는 형식(dict)으로 자동 변환
    # (최대 5개까지만 가능하므로 리스트 자르기)
    keyword_groups = [{"groupName": kw, "keywords": [kw]} for kw in keywords[:5]]

    body = {
        "startDate": start_date,
        "endDate": end_date,
        "timeUnit": time_unit,
        "keywordGroups": keyword_groups
    }

    response = requests.post(url, headers=headers, json=body)

    if response.status_code == 200:
        data = response.json()
        df_list = []

        # 🌟 [업그레이드 2] 여러 키워드의 결과를 하나의 데이터프레임으로 묶기
        for group in data['results']:
            keyword_name = group['title']

            # 해당 키워드의 검색량이 0인 기간만 있으면 data가 빈 배열일 수 있으므로 방어 코드 추가
            if 'data' in group and group['data']:
                temp_df = pd.DataFrame(group['data'])
                temp_df['키워드'] = keyword_name # 출처 라벨링
                df_list.append(temp_df)

        # 수집된 데이터가 하나라도 있으면 병합
        if df_list:
            final_df = pd.concat(df_list, ignore_index=True)
            # 컬럼 순서 및 이름 깔끔하게 정리
            final_df.rename(columns={'period': '날짜', 'ratio': '트렌드_스코어'}, inplace=True)
            final_df = final_df[['날짜', '키워드', '트렌드_스코어']] # 컬럼 순서 재배치
            return final_df
        else:
            print("데이터가 존재하지 않습니다.")
            return None

    else:
        print(f"❌ API 에러 발생: {response.status_code}\n{response.text}")
        return None

# ==========================================
# 3. 파이프라인 실행 및 검증
# ==========================================
# 비교하고 싶은 키워드를 리스트로 넣습니다. (최대 5개)
keyword_list = ["우베", "버떠떡", "두쫀쿠"]

print(f"{keyword_list} 데이터 파이프라인 가동 중...")
df_trend = get_naver_trend_pipeline(keyword_list, "2026-03-01", "2026-04-02", "date")

if df_trend is not None:
    print(f"\n✅ 파이프라인 수집 완료! 총 {len(df_trend)}행 데이터 확보.")
    print(df_trend.head())
    print("\n[데이터 샘플 확인]")
    print(df_trend.sample(5)) # 잘 섞여서 들어왔는지 무작위 확인
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 폰트 설정 (기존 환경 유지)
plt.rcParams['font.family'] = 'NanumBarunGothic' # Mac은 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

# ==========================================
# STEP 1. X축 정규화: Peak Day를 T=0 으로 맞추기
# ==========================================
# 1) 각 키워드별로 트렌드 스코어가 가장 높았던(Peak) 날짜 찾기
peak_idx = df_trend.groupby('키워드')['트렌드_스코어'].idxmax()
peak_info = df_trend.loc[peak_idx, ['키워드', '날짜']].rename(columns={'날짜': '피크_날짜'})

# 2) 원본 데이터에 피크 날짜를 붙이고, '경과 일수(T_Day)' 계산
df_norm = pd.merge(df_trend, peak_info, on='키워드', how='left')
df_norm['날짜'] = pd.to_datetime(df_norm['날짜'])
df_norm['피크_날짜'] = pd.to_datetime(df_norm['피크_날짜'])

# 핵심: 기준일(Peak)로부터 며칠 전(-)이고 며칠 후(+)인지 계산
df_norm['T_Day'] = (df_norm['날짜'] - df_norm['피크_날짜']).dt.days

# ==========================================
# STEP 2. Y축 정규화: Min-Max Scaling (0~100%)
# ==========================================
# 각 키워드별 최솟값과 최댓값 추출
min_max = df_norm.groupby('키워드')['트렌드_스코어'].agg(['min', 'max']).reset_index()
df_norm = pd.merge(df_norm, min_max, on='키워드', how='left')

# 핵심: (현재값 - 최솟값) / (최댓값 - 최솟값) * 100
df_norm['정규화_스코어(%)'] = ((df_norm['트렌드_스코어'] - df_norm['min']) /
                            (df_norm['max'] - df_norm['min'])) * 100


# ==========================================
# STEP 3. 시각화: 오버레이 및 마스터 곡선(평균) 도출
# ==========================================
plt.figure(figsize=(12, 7))

# 1) 개별 유행 아이템들의 궤적 (배경으로 얇고 투명하게 깔기)
sns.lineplot(
    data=df_norm,
    x='T_Day',
    y='정규화_스코어(%)',
    hue='키워드',
    alpha=0.3,       # 투명도 조절로 겹쳐도 보이게
    linewidth=1.5,
    legend=False     # 선이 너무 많으면 지저분하므로 범례 생략
)

# 2) 🌟 머신러닝의 정답지가 될 '마스터 평균 곡선' (두껍고 진하게)
sns.lineplot(
    data=df_norm,
    x='T_Day',
    y='정규화_스코어(%)',
    color='red',
    linewidth=4,
    label='Universal Master Curve (전체 평균 궤적)',
    errorbar=None    # 그림자(신뢰구간) 제거하여 깔끔하게
)

# 3) 차트 디테일 세팅
plt.axvline(x=0, color='black', linestyle='--', alpha=0.7, label='Peak Day (T=0)') # 중앙 수직선
plt.axhline(y=100, color='black', linestyle=':', alpha=0.5) # 천장 수평선

plt.title('유행 수명주기 가설 최종 검증 (T=0 & Min-Max 정규화 오버레이)', fontsize=16, fontweight='bold')
plt.xlabel('최고점 도달 전/후 경과 일수 (T-Day)', fontsize=12)
plt.ylabel('정규화된 검색 스코어 (Min=0, Max=100%)', fontsize=12)

# X축 범위를 -60일 ~ +60일 정도로 제한하여 '산 모양'이 잘 보이는 구간에 돋보기 대기
# (데이터 길이에 맞춰 조절하세요)
plt.xlim(-60, 60)

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. 한글 폰트 설정
plt.rcParams['font.family'] = 'NanumBarunGothic' # Mac은 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

# ==========================================
# 1. 실시간 알림 판별 함수 (EMA + Z-Score)
# ==========================================
def detect_trend_breakout(df, score_col='트렌드_스코어', ema_span=3, window=14, z_thresh=2.5, min_score=3.0):
    """
    ema_span: EMA 계산 기간 (보통 3일로 매우 민첩하게 세팅)
    window: Z-Score를 계산할 과거 비교 기간 (보통 14일~30일)
    z_thresh: 몇 표준편차를 벗어나야 알림을 울릴 것인가? (2.5면 매우 엄격함)
    min_score: 노이즈 방지용 절대 점수 하한선
    """
    d = df.copy()

    # 1) EMA(지수이동평균) 계산 - 최근 데이터에 가중치
    d['EMA'] = d[score_col].ewm(span=ema_span, adjust=False).mean()

    # 2) 과거 window 기간 동안의 평균(Mean)과 표준편차(Std) 계산
    # min_periods=1 을 주어 데이터가 부족한 극초반에도 에러가 나지 않게 함
    rolling_mean = d['EMA'].rolling(window=window, min_periods=1).mean()
    rolling_std = d['EMA'].rolling(window=window, min_periods=1).std()

    # 분모가 0이 되는 것을 방지하기 위해 아주 작은 값(1e-5) 더하기
    d['Z_Score'] = (d['EMA'] - rolling_mean) / (rolling_std + 1e-5)

    # 3) 알림(Alert) 조건 판별 (AND 조건)
    # 조건 A: 통계적 이상치 폭발 (Z-Score > 설정값)
    # 조건 B: 최소한의 검색량은 확보할 것 (EMA > 하한선)
    d['Alert'] = (d['Z_Score'] >= z_thresh) & (d['EMA'] >= min_score)

    return d

# ==========================================
# 2. 로직 검증을 위한 테스트 (우베 데이터 적용 가정)
# ==========================================
# (df_trend 데이터 중 '우베' 데이터만 뽑았다고 가정)
df_ube = df_trend[df_trend['키워드'] == '우베'].copy().reset_index(drop=True)

df_ube['날짜'] = pd.to_datetime(df_ube['날짜'])

# 함수 적용: Z-score 2.5 이상, 최소 스코어 3.0 이상일 때 알림!
df_result = detect_trend_breakout(df_ube, z_thresh=2.5, min_score=3.0)

# 알림이 울린 날짜만 추출
alert_days = df_result[df_result['Alert'] == True]

# ==========================================
# 3. 시각화: 시스템이 언제 알림을 울렸는지 확인
# ==========================================
plt.figure(figsize=(12, 5))

# 원본 데이터와 EMA 선 그리기
plt.plot(df_result['날짜'], df_result['트렌드_스코어'], label='원본 일간 스코어', color='lightgray', linestyle='--')
plt.plot(df_result['날짜'], df_result['EMA'], label='EMA (3일 지수이동평균)', color='purple', linewidth=2)

# 알림(Alert)이 터진 지점에 빨간 점 찍기
plt.scatter(alert_days['날짜'], alert_days['EMA'], color='yellow', s=80, zorder=3, label='🚨 트렌드 급상승 알림 (Triggered)')

plt.title('실시간 트렌드 경보 시스템 테스트 (EMA + Z-Score 2.5 돌파 기준)', fontsize=15, fontweight='bold')
plt.xlabel('날짜')
plt.ylabel('검색 스코어')
plt.axhline(y=3.0, color='orange', linestyle=':', alpha=0.7, label='최소 스코어 방어선 (Min=3.0)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(f"총 {len(alert_days)}번의 트렌드 급상승 알림이 발생했습니다.")
if not alert_days.empty:
    print(f"최초 발견 날짜: {alert_days['날짜'].iloc[0].strftime('%Y-%m-%d')}")
```
